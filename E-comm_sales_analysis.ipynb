# E-Commerce Sales Data Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import random
```

```python
cities = ["Delhi","Mumbai","Lucknow","Kanpur","Bangalore","Pune","Punjab","Patna"]
categories = ["Electronics","Clothing","Home","Sports"]
mon = ["Jan","Feb","Mar","Apr","May","Jun"]
data = {
    "Order_ID": range(1,1001),
    "City": [random.choice(cities) for i in range(1000)],
    "Category": [random.choice(categories) for i in range(1000)],
    "Sales": [random.randint(500,10000) for i in range(1000)],
    "Month": [random.choice(mon) for i in range(1000)]
}

df = pd.DataFrame(data)

df.head()
```

## Data Overview

```python
df.info()
df.describe()
```

## Total Sales

```python
total_sales = df["Sales"].sum()
print("Total Sales:", total_sales)
```

## Category Wise Sales

```python
category_sales = df.groupby("Category")["Sales"].sum()

print(category_sales)
```

## Category Wise Bar Chart

```python
category_sales.plot(kind="bar")

plt.title("Category Wise Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

## City Wise Sales

```python
city_sales = df.groupby("City")["Sales"].sum()

city_sales.plot(kind="pie",autopct="%1.1f%%")

plt.title("City Wise Sales Distribution")

plt.show()
```

## Month Wise Sales

```python
month_sales = df.groupby("Month")["Sales"].sum()

month_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

## Top 5 Highest Sales Orders

```python
top_orders = df.sort_values(by="Sales", ascending=False).head(5)

print(top_orders)
```

# Insights

Electronics category generated the highest revenue.

Mumbai contributed the highest sales.

Sales peak was observed in April.
